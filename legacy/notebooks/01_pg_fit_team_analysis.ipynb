설문은 자체 제작한 예시이며 검증된 척도가 아닙니다. 현재 팀 체감 응답은 새 팀 점수로 이전하지 않습니다.

```python
from pathlib import Path
import sys
root = Path.cwd() if (Path.cwd() / "src").is_dir() else Path.cwd().parent
sys.path.insert(0, str(root))
from src.config import SAMPLE_EMPLOYEES_CSV, PREFERENCE_ITEMS
from src.data_loader import load_employees_csv
from src.team_generator import generate_teams
from src.project_fit import analyze_project_candidates, rank_project_candidates, PRESETS
from src.simulator import analyze_team_candidates
from src.team_metrics import compare_member_preferences
employees = load_employees_csv(SAMPLE_EMPLOYEES_CSV)
employees.head()
```

```python
teams = list(generate_teams(employees, team_size=3))
preset = PRESETS["신규 서비스 출시"]
results = analyze_project_candidates(teams, preset["skills"], preset["roles"])
references = analyze_team_candidates(teams).drop(columns=["member_ids", "member_names"])
ranked = rank_project_candidates(results.merge(references, on="team_id", validate="one_to_one"))
ranked.head()
```

```python
best = employees[employees.employee_id.isin(ranked.iloc[0].member_ids)]
compare_member_preferences(best)
```
