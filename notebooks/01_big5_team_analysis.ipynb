# 01 Big Five Team Analysis

V1에서는 Big Five 성격 점수를 기반으로 팀 구성 후보의 성격 유사성/다양성 지표를 탐색합니다. 이 지표는 팀 성과 예측 점수가 아니라 팀 구성 특성을 설명하기 위한 분석 값입니다.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

import matplotlib.pyplot as plt
from src.config import BIG5_COLUMNS, SAMPLE_EMPLOYEES_CSV
from src.data_loader import load_employees_csv
from src.team_generator import generate_teams
from src.team_metrics import analyze_team
from src.simulator import analyze_team_candidates, get_top_diversity_teams, get_top_similarity_teams
```

## 1. 데이터 로드

```python
employees = load_employees_csv(SAMPLE_EMPLOYEES_CSV)
employees.head()
```

## 2. 직원 Big Five 데이터 확인

```python
employees[['employee_id', 'name', 'rank', 'current_department', *BIG5_COLUMNS]].head(10)
```

## 3. 기술통계

```python
employees[BIG5_COLUMNS].describe()
```

## 4. Big Five 분포 시각화

```python
employees[BIG5_COLUMNS].hist(figsize=(12, 8), bins=8)
plt.tight_layout()
plt.show()
```

## 5. 특정 팀 하나 분석

```python
sample_team = employees.head(5)
analyze_team(sample_team)
```

## 6. 팀 조합 생성

```python
teams = generate_teams(employees, team_size=5)
results = analyze_team_candidates(teams)
results.shape
```

## 7. 유사성 높은 팀 확인

```python
get_top_similarity_teams(results, top_n=5)[['team_id', 'member_ids', 'member_names', 'personality_similarity_score', 'personality_diversity_score']]
```

## 8. 다양성 높은 팀 확인

```python
get_top_diversity_teams(results, top_n=5)[['team_id', 'member_ids', 'member_names', 'personality_similarity_score', 'personality_diversity_score']]
```

## 9. 두 전략 비교

```python
comparison = results[['personality_similarity_score', 'personality_diversity_score']]
comparison.plot.scatter(x='personality_similarity_score', y='personality_diversity_score', figsize=(7, 5))
plt.title('Similarity vs Diversity Indicators')
plt.show()
```
